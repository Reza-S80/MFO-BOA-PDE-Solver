چکیده

معادلات دیفرانسیل با مشتقات جزئی غیرخطی نقش محوری در مدل‌سازی پدیده‌های پیچیده مهندسی نظیر دینامیک سیالات، انتقال حرارت، و ارتعاشات سازه‌ها دارند. با وجود گستردگی کاربرد، حل تحلیلی این معادلات در اکثر مسائل واقعی غیرممکن بوده و روش‌های عددی کلاسیک نظیر روش تفاضل محدود و روش اجزای محدود، با چالش‌هایی مانند ناپایداری عددی، وابستگی شدید به کیفیت شبکه، و پیچیدگی محاسباتی در مسائل غیرخطی مواجه‌اند.

در این پایان نامه، یک الگوریتم فراابتکاری ترکیبی نوین بر پایه الگوریتم بهینه‌سازی شعله و پروانهبرای حل عددی معادلات دیفرانسیل با مشتقات جزئی غیرخطی طراحی، پیاده‌سازی، و ارزیابی شده است. الگوریتم پیشنهادی از استراتژی ترکیب موازی با اشتراک‌گذاری اطلاعات بهره می‌برد: زیرجمعیت MFO با مکانیزم مارپیچ لگاریتمی قابلیت اکتشاف سراسری را تأمین می‌کند و زیرجمعیت BOA با مکانیزم بویایی، بهره‌برداری محلی کارآمد را فراهم می‌آورد. جواب معادله دیفرانسیل به‌صورت ترکیب خطی از توابع پایه B-spline درجه‌سوم رمزگذاری شده و تابع برازندگی بر اساس خطای باقیمانده معادله در نقاط کالوکیشین تعریف می‌گردد.

الگوریتم در محیط Python 3.10 پیاده‌سازی و روی سه معادله ارزیابی گردید: معادله گرمای خطی (برای اعتبارسنجی)، معادله برگرز (نماینده معادلات غیرخطی هذلولوی)، و معادله انتشار غیرخطی (نماینده معادلات سهموی با ضریب وابسته به جواب). نتایج حاصل از ۳۰ اجرای مستقل نشان می‌دهند که الگوریتم ترکیبی MFO-BOA با RMSE میانگینبرای معادله برگرز، به‌طور معنی‌داری از الگوریتم‌های پایه MFO مستقل  و BOA مستقل  بهتر عمل می‌کند. همچنین نرخ موفقیت ۸۶.۷٪ در مقایسه با ۵۳.۳٪ و ۴۳.۳٪ برای الگوریتم‌های پایه حاصل شده است. تحلیل حساسیت پارامتری نشان داد که اندازه جمعیت N=50 و نسبت ترکیب HR=0.5 تنظیمات بهینه برای اغلب مسائل هستند.

کد پیاده‌سازی به‌صورت متن‌باز همراه با مستندات کامل ارائه شده است تا پایه‌ای برای پایان نامه ‌های آینده در حوزه حل عددی معادلات دیفرانسیل با الگوریتم‌های فراابتکاری فراهم آورد.

کلیدواژه‌ها: معادلات دیفرانسیل با مشتقات جزئی غیرخطی؛ الگوریتم بهینه‌سازی شعله و پروانه؛ فراابتکاری ترکیبی؛ توابع پایه B-spline؛ روش‌های عددی.

پیشگفتار

مدل‌سازی ریاضی پدیده‌های پیچیده فیزیکی، شیمیایی و مکانیکی نقشی محوری در پیشرفت علم و فناوری معاصر ایفا می‌کند. معادلات دیفرانسیل با مشتقات جزئی غیرخطی ابزاری بنیادین برای توصیف این پدیده‌ها به شمار می‌روند و کاربردی گسترده در حوزه‌های مهندسی دارند: از شبیه‌سازی جریان هوا در صنعت هوافضا و مدل‌سازی موج‌های شوک در دینامیک سیالات تا تحلیل انتقال حرارت با ضریب هدایت متغیر و بررسی فرآیندهای پخش در محیط‌های متخلخل. معادلاتی چون ناویه-استوکس، برگرز، شرودینگر غیرخطی، و فیشر از جمله نمونه‌های شناخته‌شده‌ای هستند که کاربردهایی مهم در فیزیک نظری و مهندسی محاسباتی دارند.

با وجود اهمیت بالای این معادلات، یافتن جواب تحلیلی دقیق برای آن‌ها در بیشتر مسائل واقعی ممکن نیست. روش‌های عددی کلاسیک نظیر روش FDM  و روش FEM  ابزارهای اصلی دهه‌های اخیر بوده‌اند؛ اما در برخورد با مسائل شدیداً غیرخطی با محدودیت‌های جدی روبه‌رو می‌شوند: وابستگی شدید به کیفیت شبکه، پیچیدگی محاسباتی بالا در حل تکراری دستگاه‌های غیرخطی، ناپایداری عددی ناشی از محدودیت‌های گام زمانی، و دشواری تولید شبکه در هندسه‌های پیچیده. این محدودیت‌ها ضرورت جستجوی رویکردهای جایگزین و مکمل را آشکار می‌سازند.

این پایان نامه الگوریتم‌های فراابتکاری الهام‌گرفته از طبیعت را به‌عنوان رویکردی مکمل در نظر می‌گیرد. MFO  با مکانیزم مارپیچ لگاریتمی در اکتشاف سراسری قوی است، در حالی که BOA  با مکانیزم بویایی در بهره‌برداری محلی دقیق عمل می‌کند. هدف اصلی پایان نامه حاضر، طراحی، پیاده‌سازی، و ارزیابی الگوریتم ترکیبی MFO-BOA بر پایه استراتژی موازی با اشتراک‌گذاری دوره‌ای است که سینرژی این دو مکانیزم را برای حل عددی معادلات دیفرانسیل با مشتقات جزئی غیرخطی به‌کار می‌گیرد. جواب معادله با رمزگذاری به‌صورت ترکیب خطی از توابع پایه B-spline نمایش داده می‌شود و کد کامل پیاده‌سازی به‌زبان Python به‌صورت متن‌باز در پیوست ارائه شده است.

این پایان نامه سه پرسش اصلی را دنبال می‌کند: آیا الگوریتم ترکیبی MFO-BOA از الگوریتم‌های پایه و روش‌های کلاسیک عملکرد بهتری دارد؟ چگونه پارامترهای کنترلی بر رفتار همگرایی تأثیر می‌گذارند؟ و کدام استراتژی ترکیبی برای انواع مختلف معادلات کارآمدتر است؟ برای پاسخ، الگوریتم روی سه معادله آزمون در ۳۰ اجرای مستقل با الگوریتم‌های MFO، BOA، PSO، DE و روش‌های کلاسیک FDM و FEM مقایسه گردیده است. چارچوب ارزیابی بر معیارهای RMSE، MaxAE، نرخ موفقیت، انحراف معیار، و آزمون‌های آماری ویلکاکسون، فریدمن، و ANOVA استوار است.

این پایان‌نامه در پنج فصل سازماندهی شده است. فصل اول مروری جامع بر ادبیات موضوع شامل معادلات دیفرانسیل با مشتقات جزئی، روش‌های عددی کلاسیک، الگوریتم‌های فراابتکاری، و پیشینه پایان نامه  ارائه می‌دهد. فصل دوم روش‌شناسی پایان نامه را تشریح می‌کند و شامل فرمول‌بندی ریاضی دقیق، طراحی الگوریتم ترکیبی، و پروتکل ارزیابی است. فصل سوم جزئیات پیاده‌سازی، نتایج تجربی کامل، و تحلیل آماری را گزارش می‌دهد. فصل چهارم به مقایسه تطبیقی جامع، تحلیل پیچیدگی محاسباتی، و کاربردهای مهندسی الگوریتم پیشنهادی می‌پردازد. فصل پنجم نتیجه‌گیری‌ها، مشارکت‌های اصلی پایان نامه، و پیشنهادهایی برای پایان نامه ‌های آینده را ارائه می‌دهد.

## 5-1 مرورکلی پایان نامه

این پایان نامه با هدف طراحی، پیاده‌سازی و ارزیابی یک الگوریتم ترکیبی نوین بر پایه تلفیق الگوریتم بهینه‌سازی شب‌پره  و الگوریتم بهینه‌سازی پروانه  به‌منظور حل عددی معادلات دیفرانسیل با مشتقات جزئی غیرخطی صورت گرفت. انگیزه اصلی این پژوهش، شناسایی سه شکاف پژوهشی اساسی در ادبیات موضوع بود: نبود بررسی نظام‌مند ترکیب MFO-BOA برای حل ، نبود چارچوبی جامع برای ارزیابی چندمعیاری، و عدم آزمون روی معادلاتی با کاربرد مهندسی واقعی. برای پوشش این شکاف‌ها، روش‌شناسی منسجمی شامل فرمول‌بندی ریاضی دقیق مسئله، طراحی الگوریتم ترکیبی با راهبرد موازی و اشتراک‌گذاری دوره‌ای، و پروتکل ارزیابی آماری سخت‌گیرانه طراحی و اجرا شد.

## 5-2 نتیجه گیری اصلی

### 5-2-1 درباره عملکرد الگوریتم

آزمون‌های تجربی جامع بر روی سه معادله دیفرانسیل (معادله گرما، معادله برگرز و معادله انتشار غیرخطی) به نتایج زیر انجامید:

نتیجه ۱: الگوریتم MFO-BOA در هر دو مسأله غیرخطی اصلی، میانگین را نسبت به  مستقل به‌ترتیب ۶۲٪ و ۶۱٪ کاهش داد و نسبت به  مستقل ۶۹٪ و ۷۰٪ بهبود نشان داد. این برتری در سطح اطمینان ۹۹٪ (آزمون ویلکاکسون، ) معنی‌دار است.

نتیجه ۲: نرخ موفقیت MFO-BOA به‌طور میانگین در معادله برگرز ۸۶.۷٪ و در معادله انتشار غیرخطی ۸۰.۰٪ بود که به‌ترتیب ۳۳.۴ و ۴۳.۳ درصد بالاتر از بهترین الگوریتم فراابتکاری رقیب قرار دارد. این عدد بیانگر قابلیت اطمینان عملی قابل‌قبولی است.

نتیجه ۳: انحراف معیار  در ۳۰ اجرای مستقل برای  به‌طور میانگین ۴۵٪ کمتر از الگوریتم‌های پایه بود که پایداری آماری برتر این روش را تأیید می‌کند.

نتیجه ۴: روش‌های کلاسیک  و  با دقتی مشابه یا حتی بهتر، اما با سرعتی ۱۰۰۰ تا ۱۳۰۰ برابر بالاتر، برتری محاسباتی خود را حفظ می‌کنند. باید صادقانه پذیرفت که الگوریتم‌های فراابتکاری جایگزینی برای روش‌های کلاسیک در مسائل استاندارد به شمار نمی‌روند.

### 5-2-2 درباره طراحی الگوریتم

از منظر طراحی الگوریتم، نتایج حاکی از آن بود که راهبرد ترکیب موازی همراه با اشتراک‌گذاری دوره‌ای، انتخابی مناسب بوده است. این راهبرد با حفظ تنوع جمعیت و فراهم‌کردن امکان انتقال دانش، به‌طور همزمان از مزایای هر دو الگوریتم بهره می‌گیرد. تحلیل حساسیت پارامتری نشان داد که (تعادل بین دو زیرجمعیت)، (ابعاد فضای جستجو)، و  (دوره مبادله) مقادیر بهینه‌ای هستند که برای مسائل مشابه توصیه می‌شوند.

### 5-2-3 درباره کاربرد پذیری روش

مناسب‌ترین حالت‌های کاربرد الگوریتم  برای حل  عبارتند از: مسائل با هندسه‌های پیچیده نامنظم که تولید شبکه با کیفیت برای  دشوار است؛ مسائل با شرایط مرزی متغیر یا متحرک که  را ناکارآمد می‌کند؛ و مسائل چندفیزیکی پیچیده که ترکیب چند معادله دیفرانسیل با یکدیگر را می‌طلبند.

## 5-3 مشارکت های اصلی پایان نامه

این پایان نامه دستاوردهای زیر را به ادبیات موضوع افزوده است:

طراحی اولین الگوریتم ترکیبی MFO-BOA برای حل های غیرخطی: این ترکیب برای اولین بار به‌صورت سیستماتیک پیشنهاد، پیاده‌سازی، و ارزیابی شده است.

ارائه چارچوب ارزیابی جامع چندمعیاری: شامل ، نرخ موفقیت، انحراف معیار و آزمون‌های آماری دقیق را در بر می‌گیرد که می‌تواند به‌عنوان مرجعی استاندارد برای مقایسه الگوریتم‌های آتی به‌کار رود.

بررسی تجربی دقیق اثر پارامترهای الگوریتم: نقشه حساسیت پارامتری ارائه‌شده، راهنمایی کاربردی برای تنظیم الگوریتم در کاربردهای تازه فراهم می‌سازد.

کد  با اعتبارسنجی کامل: پیاده‌سازی کامل با توثیق و آزمون‌های واحد، پایه‌ای برای پایان نامه‌های آینده فراهم می‌آورد.

## 5-4 پیشنهادهایی برای پایان نامه های آینده

بر اساس یافته‌های این پایان نامه، پیشنهادهای زیر برای پژوهش‌های آتی مطرح می‌شود:

توسعه به مسائل دوبعدی و سه‌بعدی: گسترش روش به های در دامنه‌های  چالش مهمی است. در این صورت از توابع پایه دوبعدی مانند حاصل‌ضرب تانسوری توابع  یک‌بعدی استفاده می‌شود که ابعاد فضای جستجو را به  افزایش می‌دهد.

ترکیب با : تلفیق رویکرد شبکه عصبی آگاه از فیزیک با الگوریتم MFO-BOA برای آموزش پارامترهای شبکه، می‌تواند همزمان از توان بازنمایی شبکه عصبی و قابلیت اکتشاف سراسری الگوریتم ترکیبی بهره‌مند شود.

توسعه به معادلات سیستمی: بسیاری از مسائل مهندسی واقعی با سیستمی از های جفت‌شده توصیف می‌شوند. توسعه MFO-BOAبرای حل همزمان سیستم معادلات از جمله معادلات ناویه-استوکس، گام مهمی در کاربردپذیری است.

کاهش هزینه محاسباتی: استفاده از محاسبات موازی روی  برای ارزیابی همزمان کل اعضای جمعیت، قادر است زمان اجرا را تا ۱۰ تا ۵۰ برابر کاهش دهد. این توسعه، قابلیت رقابت عملی با  را برای مسائل غیرخطی پیچیده فراهم می‌آورد.

انتخاب تطبیقی توابع پایه: طراحی مکانیزمی برای انتخاب خودکار نوع و تعداد توابع پایه بر أساس ویژگی‌های جواب (مانند تعیین ناحیه‌های با گرادیان بالا) می‌تواند نیاز به تنظیم دستی را از بین ببرد.

تحلیل خطای نظری: اثبات کران‌های نظری برای خطای تقریب به‌عنوان تابعی از تعداد توابع پایه n و اندازه جمعیت N، گامی مهم در توجیه ریاضی روش خواهد بود.

کاربرد در بهینه‌سازی شکل: استفاده از MFO-BOA برای حل مسائل بهینه‌سازی شکل که در آن‌ها هم شکل دامنه و هم توزیع جواب باید بهینه شوند.

یادگیری انتقالی: بررسی امکان استفاده از جواب‌های به‌دست‌آمده برای یک پارامتر مسأله (مانند ν در معادله برگرز) به‌عنوان نقطه شروع برای مقادیر پارامتری مجاور.

ادغام با یادگیری ماشین برای یادگیری تابع هدف: برای مسائلی که ارزیابی ℱ(c) بسیار گران است، آموزش یک مدل  (مانند ) برای تقریب  می‌تواند تعداد ارزیابی‌های لازم را به‌شدت کاهش دهد.

## 5-5 کلام پایانی

این پایان نامه در مرز دو حوزه علمی پویا — الگوریتم‌های فراابتکاری و معادلات دیفرانسیل با مشتقات جزئی — قرار می‌گیرد. نتایج حاصل بیانگر آن است که تلفیق هوشمندانه الگوریتم‌های فراابتکاری با ویژگی‌های مکمل، می‌تواند عملکردی قابل‌توجه در حل PDEهای غیرخطی ارائه دهد. با این حال، این پایان نامه نقطه آغاز راه است، نه پایان آن؛ چالش‌های باقی‌مانده در حوزه مقیاس‌پذیری، تضمین همگرایی و کاربرد در مسائل واقعی مهندسی، زمینه‌ای مساعد برای پایان نامه‌های آینده هستند. امید است که چارچوب، نتایج، و کدهای متن‌باز ارائه‌شده در این پایان نامه، پایه‌ای مستحکم برای پیشرفت‌های آتی در این حوزه جذاب و پرکاربرد باشند.

# Abstract

Nonlinear Partial Differential Equations (NPDEs) play a fundamental role in modeling complex engineering phenomena such as fluid dynamics, heat transfer, and structural vibrations. Classical numerical methods including Finite Difference Method (FDM) and Finite Element Method (FEM), while widely used, face challenges including numerical instability, strong mesh dependency, and high computational complexity for strongly nonlinear problems.

This research proposes a novel hybrid metaheuristic algorithm combining Moth-Flame Optimization (MFO) and Butterfly Optimization Algorithm (BOA) for numerical solution of nonlinear PDEs. The algorithm employs a parallel hybridization strategy with information sharing: the MFO sub-population provides global exploration via logarithmic spiral mechanics, while the BOA sub-population ensures efficient local exploitation through olfactory-inspired search. Solutions are encoded as linear combinations of cubic B-spline basis functions, and the fitness function is defined based on the residual error of the governing equation at collocation points.

The algorithm was implemented in Python 3.10 and evaluated on three benchmark problems: the linear heat equation (for validation), the Burgers equation (representing nonlinear hyperbolic PDEs), and the nonlinear diffusion equation (representing parabolic PDEs with solution-dependent diffusivity). Based on 30 independent runs, MFO-BOA achieved a mean RMSE of  for Burgers’ equation, significantly outperforming standalone MFO  and BOA , with a success rate of 86.7% versus 53.3% and 43.3% respectively. Parametric sensitivity analysis identified N=50 and hybridization ratio HR=0.5 as optimal settings for most test problems.
